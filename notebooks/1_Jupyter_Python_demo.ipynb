### Functions

A self-contained block of code designed to perform a **specific task**, which you can call and reuse in different parts of your code.

A function consists of **Parameters** and **Arguments**:
- **Parameters** are variables defined in a function declaration. This act as placeholders for the values (arguments) that will be passed to the function. 
- **Arguments** are the actual values that you pass to the function when you call it. These values replace the parameters defined in the function.

The basic syntax:

- Define a function:
```python
def function_name(parameter):
     sth that you need to do here
     return the_result_you_want
```

- Call (use) a function:
```python
function_name(argument)
```

By default, a function must be called with the **correct number of arguments**. Meaning that if your function expects 2 arguments, you have to call the function with 2 arguments, not more, and not less. 

**Positional Arguments** are needed to be included in proper order, i.e the first argument is **always** listed first when the function is called, second argument needs to be called second and so on.

````python
def fun(s1,s2):
  print(s1+s2)
  
fun("Geeks","forGeeks")
````

*NB: The order of arguments must match the order of parameters in the function definition.*

**Keyword Arguments** is an argument passed to a function or method which is preceded by a keyword and equal to sign (=). 

*The order of keyword argument with respect to another keyword argument does not matter because the values are being explicitly assigned. **HOWEVER, keyword arguments should ALWAYS follow positional arguments.*** 

````python
def fun(s1, s2):
    print(s1 + s2)

# Here we are explicitly assigning the values with the positional arguments preeceding the keyword arguments
fun(s2="forGeeks", s1="Geeks")
````

A function can also return multiple values at once. For instance, we can use a commas to return them as a **tuple**.

````python
def square_and_cube(n):
    square = n ** 2
    cube = n ** 3
    return square, cube  # returns two values as a tuple

result = square_and_cube(3)
````

````python
print(f"result: {result} if of type {type(result)}")
````

But you can also return them as a **list** or **dict**.

````python
def square_and_cube(n):
    square = n ** 2
    cube = n ** 3
    return [square, cube]  # returns two values as a list

result = square_and_cube(3)
````

````python
print(f"result: {result} if of type {type(result)}")
````

Difference between what a function **does** and what it **returns**?

````python
def add_numbers_print(a, b):
    print(a + b)

result = add_numbers_print(3, 4)
print(result)
````

````python
def add_numbers_return(a, b):
    return a + b

result = add_numbers_return(3, 4)
print(result)
print(result * 2)
````

`print` → Only print value, it **cannot** be reused.

`return` → Gives only the **final value** when the function finishes, stops it at that point and sends the value back so it can be **stored** in result and **used later**.

### Lambda Function

The `lambda` keyword is used to create anonymous functions (functions without name). It is ideal for creating short, straightforward functions without extra complexity. 

`lambda` functins can process like other functions and thus return **values or lists**. 

Use a lambda function when:
- The function is a short, simple expression.
- It’s needed only once (no reuse).
- It’s used for dataframe row-wise or column-wise transformations, filtering or mapping.

Avoid a lambda function when:
- The logic needs multiple steps or statements.
- The function needs a clear name for readability.
- You plan to reuse the function.
- Situations needing debugging, docstrings, or unit tests.

````python
addition = lambda a: a + 10
addition(4)
````

Specifying a condition

````python
even_or_odd = lambda x: "Even" if x % 2 == 0 else "Odd"
even_or_odd(4)
````

### File operations

The `open()` function is a built-in function in Python that is essential for file access and manipulation. It allows us to **read**, **write**, and **append** data to a file.

````python
import pathlib
from pathlib import Path

HERE = pathlib.Path().resolve()
print(f'HERE: {HERE}')
ROOT = HERE.parent
print(f'ROOT: {ROOT}')
DATA = ROOT / 'data'
print(f'DATA: {DATA}')
````

````python
text_file_path = DATA / 'sample_text_file.txt'
text_file_path
````

````python
file = open(text_file_path, 'r') # open file in read mode
print(file.read()) # read the entire file
file.close() # close the file
````

You should always close your files, or, even better, use a `with` statement so the file automatically closes.

````python
with open(text_file_path, "r") as file:
    line = file.readline() # read one line
    print(line[:-1]) # remove the newline character at the end of the line
````

````python
with open(text_file_path, "r") as file:
    all_lines = file.readlines()
    print(all_lines)
````

Read a file row by row

````python
with open(text_file_path, "r") as file:
    for line in file.readlines():
        print(line[:-1])
````

As you can see, there are 3 different methods to read lines from a file: 

| Method       | Description                             |
|---------------|------------------------------------------|
| `read()`      | Reads the entire file as a single string |
| `readline()`  | Reads one line at a time                 |
| `readlines()` | Reads all lines and returns them as a list |

For each method, we can indicate the number of bytes from the line to return. Default is -1 to read the whole line.

Define a class and get methods and attributes

````python
class Dog:
    species = "Canis lupus familiaris"   # class attribute

    def __init__(self, name, age):     # instance attributes
        self.name = name                
        self.age = age

    def bark(self):   # method 1
        return "Woof!"

    def sit(self):   # method 2
        return "Sitting..."
    
my_dog = Dog("Buddy", 3) # Create an instance of Dog
````

````python
# define a function to get all methods of an object
def get_methods(obj):
    return [name for name in dir(obj) if callable(getattr(obj, name))] # callable allows to filter only methods

for meth in get_methods(Dog):
    print(meth)
````

````python
# define a function to get non special methods of an object (filter out dunders)
def get_methods(obj):
    return [name for name in dir(obj) if callable(getattr(obj, name)) and not name.startswith("__")] # callable allows to filter out dunders

for meth in get_methods(Dog):
    print(meth)
````

````python
# define a function to get attributes of an object
def get_attributes(obj):
    return {
        name: value
        for name, value in obj.__dict__.items()
    }

print(get_attributes(my_dog))
````

Regular methods are meant to be used by anyone, e.g., `dog.bark()`.

Methods starting with a single underscore `_` are intended for internal use by convention, e.g., `dog._growl()`.

Special methods like `__init__` or `__str__` have predefined behavior in Python, e.g., `dog = Dog("Buddy")` calls `__init__()`.

### Plotting

Matplotlib is the most popular Python library for creating static, animated, and interactive visualizations. Plotting help us understand patterns, trends, and relationships.

It graphs your data on **Figures** (e.g., windows, Jupyter widgets, etc.), each of which can contain one or more `Axes`, an area where points can be specified in terms of *x-y coordinates* (or theta-r in a polar plot, x-y-z in a 3D plot, etc.). 

An ***Axes*** is the **area where your data are drawn** attached to a Figure using `Axes.plot`, usually containing 2 objects (or 3 for 3D) with **ticks** and **tick labels** to provide scales. Each Axes can also have a **title** (`set_title()`), an **x-label** (`set_xlabel()`), and a **y-label** (`set_ylabel()`).

***Axis*** are **objects that set the scale and limits and generate ticks** (the marks on the Axis) and **ticklabels** (strings labeling the ticks). The location of the ticks is determined by a `Locator` object and the ticklabel strings are formatted by a `Formatter`. The combination of the correct Locator and Formatter gives very fine control over the tick locations and labels.
