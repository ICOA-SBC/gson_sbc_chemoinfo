## Exercise 3: functions

```python
# Define a function calculation, which takes x as parameter and return ((5 * x) / 2)²
def calculation(x):
  return ((5 * x) / 2) ** 2
```

```python
# Verify the result of calculation(4)
calculation(4) == 100
```

```python
# Define the same function f using a lambda function
f = lambda x: ((5 * x) / 2) ** 2

# print f(4)
print(f(4))
```
