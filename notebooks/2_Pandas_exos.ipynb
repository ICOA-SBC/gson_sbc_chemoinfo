## Exercise 1: Create and update your own dataframe

### Questions

Using the provided data, create a pandas DataFrame.

| Name  | math score | physics score |
| ----- | ---------- | ------------ |
| Tom   | 20         | 15           |
| Lucie | 17         | 13           |
| Emma  | 4          | 16           |
| Lucas | 12         | 2            |

```python
data = {
  'Name': ['Tom', 'Lucie', 'Emma', 'Lucas'],
  "math score": [20, 17, 4, 12],
  "physics score": [15, 13, 16, 2]
}

df_dic = pd.DataFrame(data)
df_dic
```

You are given a dictionary that assigns letter grades to ranges of scores.

```python
# Given grade mapping
grade = {
    'A': [20,19,18,17], 
    'B': [16,15,14,13], 
    'C': [12,11,10], 
    'D': [9,8,7,6], 
    'E': [5,4,3], 
    'F': [2,1,0]
}

grade
```

Create a function called `get_grade` that takes a numeric score as input and returns the corresponding grade.

```python
def get_grade(score):
    for g, values in grade.items():
        if score in values:
            return g
    return None
```

```python
# Test your function
get_grade(15) == 'B'
```

Then, `map`or `apply` this function to the DataFrame to create two new columns: one for math grades (`math_grade`) and one for physics grades (`physics_grade`). Finally, display the updated DataFrame with the new grade columns.

```python
# Apply function to create new column
df_dic['math_grade'] = df_dic['math score'].map(get_grade)
df_dic['physics_grade'] = df_dic['physics score'].apply(get_grade)
df_dic
```

Presence of 100 NaN values for `math score` and 50 for `reading score`. 

Missing values appear when some entries in a dataset are left blank, marked as NaN, None or special strings like "?". If not handled properly, they can reduce accuracy, create bias and break algorithms that require complete data. There are several strategy to handle missing values, dependending on the amound of data and/or the goal. The most common ones include:
1. Simply dropping the entire rows containing those missing values.
2. Imputing missing values with mean/median.

```python
# Display the number of NaN values in each column
df.isnull().sum()
```

```python
df.isna().sum()
```

```python
df.isnull?
```

```python
# if NaN values exist, replace by the median of the column using fillna()
df['math score'] = df['math score'].fillna(df['math score'].median())
df['reading score'] = df['reading score'].fillna(df['reading score'].median())
```

## Exercise 3: Failed or Passed?

Now we want to know which students effectively pass their exams or not. To do so, we will set the **minimum marks to 40** to pass the exam.

```python
# Set a minimum passing mark
passmark = 40
```

Let's see for math exam

### Questions

Write a command to create a new boolean column `Math_PassStatus` to indicate if the student passed or failed in math.

```python
# 1st method
df['Math_PassStatus'] = df['math score'] >= passmark # simpler way to create a boolean column
```

```python
# 2nd method
df['Math_PassStatus'] = df['math score'].map(lambda x: 'Pass' if x >= passmark else 'Fail') # another way to create a categorical column using apply and lambda
```

```python
# Check the dataframe
df.head()
```

### Questions

How many students passed their math ?

```python
df['Math_PassStatus'].value_counts()
```

Now check if parental level of education could influence students' math results by computing mean and median for each group of parental level of education.

```python
# mean
df.groupby("parental level of education")["math score"].mean().round(1)
```

```python
# median
df.groupby("parental level of education")["math score"].median()
```
