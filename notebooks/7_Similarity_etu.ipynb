How to transform human non readable explicit bitvectors in readable bitstrings ?

```python
# List attributes and methods of the object maccs_fp0 to select the right method
XXX(maccs_fp0)
```

```python
# convert the explicit bitvector in a bitstring (bs0) and print it
bs0 = maccs_fp0.XXX
print(bs0)
```

```python
# how long is the bitstring ?
print(f'Length of the bitstring: {XXX(bs0)}, not 166')
```

167, not 166, explanation:

- The original MACCS keys are defined as 166 keys, numbered 1 to 166
- RDKit adds a bit 0, which is always set to 0: this gives 167 bits total, but only bits 1–166 are chemically meaningful

Why does RDKit do this?

- Indexing compatibility
- MACCS keys were historically 1-indexed
- Programming languages like C++ and Python are 0-indexed

Keeping bit 0 as a dummy bit allows MACCS key i to map directly to bit i. It makes it easier to compare RDKit MACCS keys with the original documentation.
